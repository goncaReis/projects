```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
data_folder_path = Path('../data').resolve()
```

```python
#turn each csv into pandas dataframe
df_customers = pd.read_csv( data_folder_path / 'olist_customers_dataset.csv')
df_order_reviews = pd.read_csv( data_folder_path / 'olist_order_reviews_dataset.csv')
df_orders = pd.read_csv( data_folder_path / 'olist_orders_dataset.csv')
df_order_items = pd.read_csv( data_folder_path / 'olist_order_items_dataset.csv')
df_order_payments = pd.read_csv( data_folder_path / 'olist_order_payments_dataset.csv')
df_geolocation = pd.read_csv( data_folder_path / 'olist_geolocation_dataset.csv')
df_resellers = pd.read_csv( data_folder_path / 'olist_resellers_dataset.csv')
df_products = pd.read_csv( data_folder_path / 'olist_products_dataset.csv')
```

### DATA EXPLORATION

```python
#let's first look at the orders table by checking how many rows and columns it has
#what are the columns and the datatypes
#number of null values
#number of duplicates
#and looking at the first 3 rows to get a good graps of the data inside

print(f'The table has {df_orders.shape[0]} rows and {df_order_items.shape[1]} columns.\n')
print(f'{df_orders.dtypes}\n')
print(f'{df_orders.info()}\n')
df_orders_duplicates = df_orders.duplicated().sum()
print(f'The table has {df_orders_duplicates} duplicates.\n')
df_orders.head(3)
```

```python
#cast date columns from string to pd.datetime in order to perform time-based calculations
df_orders['order_purchase_timestamp'] = pd.to_datetime(df_orders['order_purchase_timestamp'])
df_orders['order_approved_at'] = pd.to_datetime(df_orders['order_approved_at'])
df_orders['order_delivered_carrier_date'] = pd.to_datetime(df_orders['order_delivered_carrier_date'])
df_orders['order_delivered_customer_date'] = pd.to_datetime(df_orders['order_delivered_customer_date'])
df_orders['order_estimated_delivery_date'] = pd.to_datetime(df_orders['order_estimated_delivery_date'])

#validate if the datatypes are now datetime
df_orders.dtypes
```

```python
print(f'The table has {df_order_items.shape[0]} rows and {df_order_items.shape[1]} columns.\n')
print(f'{df_order_items.dtypes}\n')
print(f'{df_order_items.info()}\n')
df_order_items_duplicates = df_order_items.duplicated().sum()
print(f'The table has {df_order_items_duplicates} duplicates.\n')
print(f'{df_order_items.describe()}\n')
df_order_items.head(3)
```

```python
#how many unique orders have there been?
unique_orders = df_order_items['order_id'].nunique()

#how many unique products have been sold?
unique_products = df_order_items['product_id'].nunique()

print(f' There have been {unique_orders} orders.')
print(f' There have been {unique_products} products.')
```

```python
fig, (ax_box, ax_hist) = plt.subplots(
    2,
    figsize=(10, 6),
    sharex=True
)

# Boxplot
ax_box.boxplot(df_order_items["price"], vert=False)
ax_box.set_title("Price Distribution")

# Histogram
ax_hist.hist(df_order_items["price"], bins=50)
ax_hist.set_xlabel("Price (BRL)")
ax_hist.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
