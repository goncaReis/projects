## Olist E-commerce Dataset: Project Overview

### Dataset Description

This project analyzes the Olist Brazilian E-commerce Public Dataset, a real-world transactional dataset containing information from orders placed on the Olist marketplace between 2016 and 2018. The dataset consists of multiple relational tables that capture different aspects of the customer journey, including customers, orders, products, sellers, payments, reviews, and geolocation.

The data represents a complete e-commerce ecosystem, allowing analysis from multiple perspectives, such as sales performance, customer behavior, logistics, payment methods, seller performance, and customer satisfaction.

By combining information across these related tables, it is possible to reconstruct the lifecycle of an order—from purchase and payment to delivery and customer review—and derive actionable business insights.

### Business Context

Olist is a Brazilian e-commerce platform that connects small and medium-sized merchants to major online marketplaces. As a marketplace intermediary, Olist manages relationships between customers, sellers, logistics providers, and payment systems.

For a business operating at this scale, data plays a critical role in monitoring operational efficiency and customer experience. Questions such as:

Which product categories generate the most revenue?
Which sellers contribute the most to overall sales?
How long do deliveries take across different regions?
What factors influence customer review scores?
Which payment methods are most commonly used?
Are there geographic differences in purchasing behavior?

are essential for making informed business decisions.

This exploratory data analysis aims to investigate these questions by examining the relationships between the dataset's tables and identifying patterns that could support strategic improvements.

### Project Objectives

The objectives of this project are to:
